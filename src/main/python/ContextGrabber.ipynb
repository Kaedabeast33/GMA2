````python

'''
? create handler that picks from an array of options for which group 

3 create client data processor agent class

2 create Gma Context grabber + shows navigation reasoning (all columns with  tags and descriptions and table and tags and descriptions , all columns in specified ma, all columns in specified table)

create Gma db






'''

class GmaContextAgent:
    def __init__(self,chat=None):
        self.instructions = ''
        self.gmaContextObj = GmaContextObj(chat)
    

    
        

class ContextObj:
    def __init__(self, context_dict=None):
        context_dict = context_dict or {}

        self.mas = context_dict.get("mas", [])
        self.tables = context_dict.get("tables", [])
        self.columns = context_dict.get("columns", [])




class DbWhereObj:
    def __init__(self,chat):
        self.m = ''
    def to_dict(self):
        return {}
        

class DataObj:
    def __init__(self):
        self.m = ''





class GmaContextObj:
    def __init__(self, chat= None,contextObj = None):
    # Store the arrays directly from the JSON structure    
        if chat is None:
            # Use provided context or blank
            self.contextObj = contextObj if contextObj is not None else ContextObj()
        else:
            # AI-generated context
            self.contextObj = GmaContextObj.createContextObjFromChat(chat)
        
        
        
    def to_dict(self):
        return{
            "mas":self.contextObj.mas,
            "tables":self.contextObj.tables,
            "columns":self.contextObj.columns,
        }

    @staticmethod
    def createContextObjFromChat(chat):
        with open('contextParserPrompt.txt', 'r', encoding='utf-8') as f:
            system_prompt = f.read()
        
        try:
            # Make API call
            resp = client.messages.create(
                model="claude-sonnet-4-20250514",
                max_tokens=4096,
                temperature=0,
                system=system_prompt,
                messages=[{"role": "user", "content": chat}]
            )
            
            # Extract response text
            response_text = resp.content[0].text.strip()
            
            # Remove markdown code blocks if present
            if response_text.startswith('```'):
                lines = response_text.split('\n')
                response_text = '\n'.join(lines[1:-1]) if len(lines) > 2 else response_text
                response_text = response_text.replace('```json', '').replace('```', '').strip()
            
            # Parse JSON response
            context_json = json.loads(response_text)
            
            # Create GmaContextObj with the parsed JSON
            print(context_json)
            
            print("Successfully parsed context object")
            return ContextObj(context_json)
            
        except json.JSONDecodeError as e:
            print(f"Error parsing JSON: {e}")
            print(f"Raw response: {response_text}")
            return None
        except Exception as e:
            print(f"API Error: {e}")
            return None

            




# Access the data




# Usage:
g = GmaContextObj('''
I need data from my halloween ma I also need data from my hamburger ma and the tables condiments and george only from that ma
''')


print("MAs:", g.contextObj.mas)
print("Tables:", g.contextObj.tables)
print("Columns:", g.contextObj.columns)

        
        






        




'''




'''



        
        
    
````
